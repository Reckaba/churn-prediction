```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/telco_churn.csv')
```

```python
print("Размер датасета:", df.shape)
print("\nПервые 5 строк:")
df.head()
```

```python
print("Типы данных:")
print(df.dtypes)
print("\nПропуски:")
print(df.isnull().sum())
print("\nРаспределение Churn:")
print(df['Churn'].value_counts())
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Churn', palette='Set2')
plt.title('Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Count')
plt.savefig('../data/churn_dist.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Contract', hue='Churn', palette='Set2')
plt.title('Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Count')
plt.savefig('../data/churn_by_contract.png')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='tenure', hue='Churn', bins=30, palette='Set2')
plt.title('Churn by Tenure')
plt.xlabel('Tenure (months)')
plt.ylabel('Count')
plt.savefig('../data/churn_by_tenure.png')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='MonthlyCharges', hue='Churn', bins=30, palette='Set2')
plt.title('Churn by Monthly Charges')
plt.xlabel('Monthly Charges ($)')
plt.ylabel('Count')
plt.savefig('../data/churn_by_charges.png')
plt.show()
```
